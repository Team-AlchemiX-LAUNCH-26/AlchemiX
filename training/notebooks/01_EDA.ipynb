# 01 EDA

Explore the supplied traffic, telemetry, and incident datasets before training.

```python
from pathlib import Path
import pandas as pd

ROOT = Path.cwd().resolve().parents[1]
DATA = ROOT / 'datasets'

traffic = pd.read_csv(DATA / 'link_traffic_history.csv')
telemetry = pd.read_csv(DATA / 'link_telemetry.csv')
incidents = pd.read_csv(DATA / 'link_incident_history.csv')
```

```python
summary = pd.DataFrame({
    'dataset': ['traffic', 'telemetry', 'incidents'],
    'rows': [len(traffic), len(telemetry), len(incidents)],
    'links': [traffic.link_id.nunique(), telemetry.link_id.nunique(), incidents.link_id.nunique()],
    'tick_min': [traffic.tick.min(), telemetry.tick.min(), incidents.tick.min()],
    'tick_max': [traffic.tick.max(), telemetry.tick.max(), incidents.tick.max()],
})
display(summary)
display(traffic.head())
display(telemetry.head())
display(incidents.head())
```

```python
eda_checks = {
    'saturated_traffic_rows': int((traffic['status'] == 'saturated').sum()),
    'max_load_ratio': float(traffic['load_ratio'].max()),
    'missing_observed_latency': int(traffic['observed_latency_ms'].isna().sum()),
    'under_reporting_20pct_rows': int(((telemetry['measured_latency_ms'] - telemetry['self_reported_latency_ms']) / telemetry['measured_latency_ms'] >= 0.20).sum()),
    'jammed_rows': int(incidents['jammed_flag'].sum()),
    'jam_rate': float(incidents['jammed_flag'].mean()),
}
pd.Series(eda_checks)
```
