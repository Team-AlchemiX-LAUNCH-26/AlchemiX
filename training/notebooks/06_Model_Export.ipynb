# 06 Model Export

Verify exported artifacts, feature order, and manifest metadata consumed by the Go runtime.

```python
from pathlib import Path
import json
import pandas as pd

ROOT = Path.cwd().resolve().parents[1]
MODELS = ROOT / 'internal' / 'models'

artifact_names = ['congestion_model.json', 'trust_model.json', 'targeting_model.json', 'model_manifest.json']
artifacts = []
for name in artifact_names:
    path = MODELS / name
    artifacts.append({'artifact': name, 'exists': path.exists(), 'size_bytes': path.stat().st_size if path.exists() else 0})
pd.DataFrame(artifacts)
```

```python
for name in ['congestion_model.json', 'trust_model.json', 'targeting_model.json']:
    model = json.loads((MODELS / name).read_text())
    print('\n', name)
    print('task:', model.get('task'))
    print('version:', model.get('model_version'))
    print('feature_count:', len(model.get('feature_names', [])))
    print('dataset_hash:', model.get('dataset_hash'))
```

```python
manifest = json.loads((MODELS / 'model_manifest.json').read_text())
display(manifest['split_strategy'])
display({name: data['artifact'] for name, data in manifest['models'].items()})
```
