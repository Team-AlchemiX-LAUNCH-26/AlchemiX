# 04 Model Training

Run the reproducible training scripts. The scripts export Go-compatible JSON artifacts under `internal/models/`.

```python
from pathlib import Path
import subprocess
import sys

ROOT = Path.cwd().resolve().parents[1]
TRAINING = ROOT / 'training'

scripts = ['validate_data.py', 'train_congestion.py', 'train_trust.py', 'train_targeting.py']
for script in scripts:
    print(f'Running {script}')
    subprocess.run([sys.executable, str(TRAINING / script)], cwd=ROOT, check=True)
```

```python
models_dir = ROOT / 'internal' / 'models'
for path in sorted(models_dir.glob('*_model.json')):
    print(path.name, path.stat().st_size, 'bytes')
```
