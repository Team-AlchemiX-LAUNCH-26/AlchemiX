# 05 Model Evaluation

Load exported model metadata, inspect champion metrics, and generate consolidated reports.

```python
from pathlib import Path
import json
import subprocess
import sys
import pandas as pd

ROOT = Path.cwd().resolve().parents[1]
TRAINING = ROOT / 'training'
MODELS = ROOT / 'internal' / 'models'

subprocess.run([sys.executable, str(TRAINING / 'evaluate_all.py')], cwd=ROOT, check=True)
```

```python
manifest = json.loads((MODELS / 'model_manifest.json').read_text())
rows = []
for name, data in manifest['models'].items():
    rows.append({
        'model': name,
        'task': data['task'],
        'selected_model': data['model_type'],
        'feature_count': len(data['features']),
        'dataset_hash': data['dataset_hash'],
    })
display(pd.DataFrame(rows))
```

```python
for name, data in manifest['models'].items():
    print('\n', name.upper())
    display(pd.Series(data['validation_metrics']))
```
